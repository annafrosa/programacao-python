## Exercício 1 (Fácil) - Medição de tensão

Um técnico realizou 5 medições de tensão em um circuito. Os valores medidos foram armazenados em uma lista:

tensoes = [218, 220, 221, 219, 223]

Objetivo: escreva um programa que percorra a lista e mostre cada tensão na tela.

```python
# Resposta

tensoes = [218, 220, 221, 219, 223]

for tensao in tensoes:
    print(f"Tensão: {tensao} V")
```

## Exercício 2 (Médio) — Análise de tensão

Um técnico deseja analisar se as tensões medidas em um circuito estão dentro do intervalo considerado normal. 

Primeiro, o programa deve perguntar ao usuário quantas medições serão realizadas. Em seguida, deve ler cada valor de tensão e deve verificar se ela está normal ou fora do limite.

Considere que a tensão está normal quando estiver entre 210 V e 230 V.

Objetivo: Escreva um programa que percorra a lista e mostre cada tensão na tela.

```python
medicoes = int(input("Quantidade de medicoes: "))

for i in range(1, medicoes + 1):
    tensao = float(input(f"Digite a tensão: "))

    if tensao >= 210 and tensao <= 230:
        print(f"{tensao} V - Normal")
    else:
        print(f"{tensao} V - Fora do limite")
```

## Exercício 3 (Difícil) - Análise de Corrente

Um estudante possui diferentes resistores e deseja verificar qual corrente passaria por cada um deles quando conectados a uma fonte de 12 V. Os valores das resistências devem ser informados pelo usuário e armazenados em uma lista. 

Para cada resistor, calcule a corrente utilizando a Lei de Ohm:

I = V/R


```python
num_resistores = int(input("Quantidade de resistores: "))

# A fonte informada é de 12V, então a tensão V é constante para todos os resistores.
V = 12 
resistencias = []

for i in range(num_resistores):
    resistencia = float(input(f"Digite a resistência: "))
    resistencias.append(resistencia)

for resistencia in resistencias:
    I = V / resistencia

    if I > 1:
        print(f"R: {resistencia} ohm | I: {I} A | Alta")
    else:
        print(f"R: {resistencia} ohm | I: {I} A | Normal")
```
