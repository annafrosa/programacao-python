# Desafio — Funções e gráficos de sinais

Em sistemas elétricos, é comum analisar diferentes sinais no domínio
do tempo. Neste desafio, você deverá criar funções em Python para
calcular e plotar sinais de tensão senoidal e quadrada utilizando as
bibliotecas **numpy** e **matplotlib**.

## Dados do problema

Considere os seguintes valores:

- Amplitude (Vm): 5 V
- Frequência (f): 60 Hz
- Intervalo de tempo: de 0 a 0,04 s
- Número de amostras: 401 pontos

### Objetivo

O programa deve:

1. Criar uma função para calcular a tensão senoidal:
   
   $$V_s(t) = V_m \cdot \sin(2\pi f t)$$

2. Criar uma função para calcular a tensão quadrada.

   A tensão quadrada deve assumir o valor $+V_m$ quando $\sin(2\pi f t) \geq 0$ e o valor $-V_m$ quando $\sin(2\pi f t) < 0$.

3. Gerar o vetor de tempo entre 0 e 0,04 s usando `np.linspace()`.
4. Calcular os dois sinais utilizando as funções criadas.
5. Plotar os dois sinais no mesmo gráfico.
6. Adicionar título, legenda e nomes aos eixos do gráfico.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Funcao para calcular o sinal senoidal
def sinal_senoidal(amplitude, frequencia, tempo):
    return amplitude * np.sin(2 * np.pi * frequencia * tempo)

def sinal_quadrado(amplitude, frequencia, tempo):
    seno = np.sin(2 * np.pi * frequencia * tempo)

    return np.where(seno >= 0, amplitude, -amplitude)

Vm = 5  # Amplitude do sinal
f = 60 # Frequência do sinal
t = np.linspace(0, 0.04, 401)  # Vetor de tempo

# Calculo dos sinais
s_senoidal = sinal_senoidal(Vm, f, t)
s_quadrado = sinal_quadrado(Vm, f, t)

# Grafico
plt.plot(t, s_senoidal, label='Sinal Senoidal', color='blue')
plt.plot(t, s_quadrado, label='Sinal Quadrado', color='red')
plt.title('Sinais Senoidal e Quadrado')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude (V)')
plt.grid()
plt.legend()
plt.show()
```
