## 1. O que são imports?

Python possui muitas funcionalidades prontas. Algumas estão disponíveis diretamente, como `print()`, `len()` e `sum()`.

Outras funcionalidades ficam organizadas em **módulos e bibliotecas**. Para utilizá-las, precisamos fazer um **import**.

Por exemplo:

```python
import math
```

Depois disso, podemos utilizar funções e constantes do módulo `math`.

## 2. Biblioteca padrão x biblioteca externa

### Biblioteca padrão

São módulos que já fazem parte da instalação do Python.

Exemplos:

### Bibliotecas externas

## 3. Importando o módulo `math`

A forma mais simples é:

```python
import math
```

Depois, usamos o nome do módulo seguido de `.` e do recurso que queremos utilizar.

````python
import math

print(math.pi)
print(math.sqrt(25))
````

## 4. Funções importantes do `math`

Algumas funções que serão úteis em Engenharia:

| Função | Significado |
|---|---|
| `math.sqrt(x)` | raiz quadrada |
| `math.pow(x, y)` | potência |
| `math.sin(x)` | seno |
| `math.cos(x)` | cosseno |
| `math.tan(x)` | tangente |
| `math.log(x)` | logaritmo natural |
| `math.log10(x)` | logaritmo na base 10 |
| `math.ceil(x)` | arredonda para cima |
| `math.floor(x)` | arredonda para baixo |

O módulo também possui constantes, como `math.pi` e `math.e`.

````python
import math

print("Raiz de 81:", math.sqrt(81))
print("2 elevado a 5:", math.pow(2, 5))
print("PI:", math.pi)
print("e:", math.e)
````

## 5. Graus e radianos

As funções trigonométricas do `math` trabalham com **radianos**.

Para converter graus para radianos:

```python
math.radians(angulo)
```

Para converter radianos para graus:

```python
math.degrees(angulo)
```

````python
import math

angulo_graus = 90
angulo_rad = math.radians(angulo_graus)

print("90 graus em radianos:", angulo_rad)
print("sen(90°):", math.sin(angulo_rad))
print("cos(90°):", math.cos(angulo_rad))
````

## 6. Outra forma de importar: `from ... import`

Em vez de escrever `math.sqrt()`, podemos importar diretamente uma função:

```python
from math import sqrt
```

Assim:

```python
sqrt(25)
```

Também podemos importar vários recursos:

````python
from math import sqrt, pi

print(sqrt(36))
print(pi)
````

## 7. Usando `as` para criar um apelido

Também podemos dar um apelido para um módulo.

Por exemplo:

```python
import math as m
```

Agora podemos escrever `m.sqrt()` em vez de `math.sqrt()`.

````python
import math as m

print(m.sqrt(49))
print(m.pi)
````

## 8. Introdução ao `matplotlib`

O `matplotlib` é uma biblioteca muito utilizada para criar gráficos em Python.

Um dos módulos mais usados é o `pyplot`.

Por convenção, utilizamos o apelido `plt`:

```python
import matplotlib.pyplot as plt
```

O `as plt` não é obrigatório, mas é uma convenção muito comum.

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.show()
````

## 9. Elementos básicos de um gráfico

Alguns comandos importantes:

```python
plt.plot(x, y)
plt.title("Título")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.legend()
plt.grid()
plt.show()
```

Esses comandos permitem transformar um gráfico simples em uma visualização adequada para apresentar resultados.

````python
import matplotlib.pyplot as plt

x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]

plt.plot(x, y, label="y = x²")
plt.title("Função quadrática")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
````

## 10. Gráficos de funções

Podemos combinar funções que aprendemos na aula anterior com o `matplotlib`.

Por exemplo, podemos criar uma função que calcula `y = x²` e depois representar seus valores graficamente.

````python
import matplotlib.pyplot as plt

def calcular_y(x):
    return x ** 2

x = list(range(-5, 6))
y = []

for valor in x:
    y.append(calcular_y(valor))

plt.plot(x, y, label="y = x²")
plt.title("Gráfico de uma função")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
````

## 11. Exemplo de Engenharia: sinal senoidal

Uma aplicação importante é representar sinais.

Um sinal senoidal pode ser representado por:

$$x(t) = A\sin(2\pi f t)$$

onde:

- `A` é a amplitude;
- `f` é a frequência em Hz;
- `t` é o tempo em segundos.

Neste exemplo, vamos utilizar `numpy` para gerar a sequência de tempo e calcular o seno, e `matplotlib` para visualizar o sinal.

### O que é o `numpy`?

O `numpy` é uma biblioteca externa muito usada em ciência de dados, engenharia, física e machine learning.

A principal ideia do `numpy` é permitir que a gente trabalhe com **arrays** (vetores e matrizes) de forma rápida e eficiente.

Em vez de percorrer elemento por elemento com `for`, o `numpy` consegue aplicar operações matemáticas em toda a sequência de uma vez.

Exemplo:

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = x * 2
print(y)
```

Resultado:

```python
[ 2  4  6  8 10]
```

Isso é muito útil para sinais, gráficos, processamento de dados e muitos cálculos de Engenharia.

````python
import numpy as np
import matplotlib.pyplot as plt

amplitude = 1
frequencia = 5
tempo = np.linspace(0, 1, 1000)
sinal = amplitude * np.sin(2 * np.pi * frequencia * tempo)

plt.plot(tempo, sinal, label="Sinal senoidal")
plt.title("Sinal Senoidal")
plt.xlabel("Tempo (s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()
````

## 12. Comparando dois sinais

O `matplotlib` também permite colocar mais de uma curva no mesmo gráfico.

Isso será muito importante posteriormente para comparar sinais em aplicações de Engenharia.

````python
import numpy as np
import matplotlib.pyplot as plt

tempo = np.linspace(0, 1, 1000)
sinal_1 = np.sin(2 * np.pi * 5 * tempo)
sinal_2 = np.sin(2 * np.pi * 10 * tempo)

plt.plot(tempo, sinal_1, label="5 Hz")
plt.plot(tempo, sinal_2, label="10 Hz")
plt.title("Comparação entre sinais")
plt.xlabel("Tempo (s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()
````

## 13. Organização: imports no início do código

Uma boa prática é colocar os imports no início do programa.

Exemplo:

```python
import numpy as np
import matplotlib.pyplot as plt

# cria uma sequência de tempo
tempo = np.linspace(0, 1, 1000)

# calcula um sinal senoidal
sinal = np.sin(2 * np.pi * 5 * tempo)

def calcular_sinal(...):
    ...

# restante do programa
```

Isso facilita a leitura e deixa claro quais bibliotecas o programa utiliza.

O `numpy` é uma biblioteca externa muito importante para computação numérica. Ele permite trabalhar com arrays e aplicar operações matemáticas em vários valores ao mesmo tempo. Por exemplo, `np.linspace()` cria uma sequência de números e `np.sin()` calcula o seno de todos eles em uma única operação.

Já o `math` é uma biblioteca interna da linguagem Python. Ele também serve para cálculos matemáticos, mas é mais indicado para valores individuais, enquanto o `numpy` é mais prático para dados em série, como sinais, listas de medidas e gráficos de engenharia.

1. a área;

Utilize `np.pi`.

Fórmulas:

# Desafio: preparando para a próxima aula

Crie uma função chamada `gerar_sinal(frequencia, amplitude)`.

A função deve gerar um sinal senoidal e retornar os valores de tempo e amplitude.

Depois, utilize `matplotlib` para visualizar o sinal.

Teste com:

- frequência = 5 Hz e amplitude = 1;
- frequência = 10 Hz e amplitude = 2.

Esse exercício combina três conceitos importantes:

**funções + imports + gráficos**.

````python
import numpy as np
import matplotlib.pyplot as plt

def gerar_sinal(frequencia, amplitude, tempo_total=1, pontos=1000):
    tempo = np.linspace(0, tempo_total, pontos)
    sinal = amplitude * np.sin(2 * np.pi * frequencia * tempo)
    return tempo, sinal

tempo_1, sinal_1 = gerar_sinal(5, 1)
tempo_2, sinal_2 = gerar_sinal(10, 2)

plt.plot(tempo_1, sinal_1, label="5 Hz e amplitude 1")
plt.plot(tempo_2, sinal_2, label="10 Hz e amplitude 2")
plt.title("Sinais senoidais")
plt.xlabel("Tempo (s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()
````
