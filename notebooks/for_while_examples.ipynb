# Laços de repetição em Python

Vamos aprender duas formas de repetir instruções: `for` e `while`.

Objetivos:
- identificar quando usar cada estrutura;
- entender a diferença entre repetir "um número fixo de vezes" e "enquanto uma condição for verdadeira";
- praticar com exemplos simples e seguros.

## 1) `for`: repetir um número conhecido de vezes

Use `for` quando você sabe quantas vezes quer repetir ou quando quer percorrer itens de uma sequência.

`range(inicio, fim)` vai de `inicio` até `fim - 1`.

```python
for numero in range(1, 6):
    print(numero)
```

### Observação importante

`range(1, 6)` produz: 1, 2, 3, 4, 5.
O valor final (`6`) não é incluído.

Também podemos usar `range(0, 11, 2)` para imprimir números pares.

```python
for numero in range(0, 11, 2):
    print(numero)
```

```python
notas = [7.5, 8.0, 9.2, 6.8]
soma = 0

for nota in notas:
    soma = soma + nota

media = soma / len(notas)
print(f"Média: {media:.2f}")
```

### Exemplo: `for` com `if` e `else`

Podemos usar `if` e `else` dentro de um `for` para processar cada item da lista de forma diferente.

```python
# Verificando se notas estão aprovadas
for nota in notas:
    if nota >= 7:
        print(f"Nota {nota} - APROVADO")
    else:
        print(f"Nota {nota} - REPROVADO")
```

## 2) `while`: repetir enquanto a condição for verdadeira

Use `while` quando a repetição depende de uma condição.

Atenção: a condição precisa mudar ao longo do programa. Se ela nunca mudar, o código pode entrar em um loop infinito.

```python
# Qual erro nesse código?

numero = 1

while numero <= 5:
    print(numero)
```

```python
# Correção

numero = 1

while numero <= 5:
    print(numero)
    numero = numero + 1
```

## Comparando `for` e `while`

- `for`: melhor quando o número de repetições é conhecido ou quando queremos percorrer elementos de uma sequência.
- `while`: melhor quando a repetição depende de uma condição.

Em outras palavras:
- `for` = "faça isso N vezes"
- `while` = "faça isso enquanto a condição for verdadeira"

```python
numero = int(input("Digite um número (0 para sair): "))

while numero != 0:
    print(f"Você digitou: {numero}")
    numero = int(input("Digite um número (0 para sair): "))

print("Programa encerrado.")
```

```python
# Dica avançada: o comando break encerra o laço
numero = 0

while True:
    numero = int(input("Digite um número (0 para sair): "))
    if numero == 0:
        break
    print(f"Você digitou: {numero}")

print("Programa encerrado.")
```

2. Use `while` para somar os números de 1 até 5.
3. Explique em voz alta: quando você usaria `for` e quando usaria `while`.

Dica: pense em "quantidade conhecida" versus "condição".

```python
# Exercicio final 1: imprimir os números pares de 2 a 10 usando um laço for

for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```

```python
# Exercício final 2: soma dos números de 1 até 5
soma = 0
numero = 1

while numero <= 5:
    soma = soma + numero
    numero = numero + 1

print("Soma:", soma)
```
