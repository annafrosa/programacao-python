## 1. O que é uma função?

Uma função é um **bloco de código criado para realizar uma tarefa específica**.

Em vez de escrever o mesmo código várias vezes, podemos colocar esse código dentro de uma função e chamá-la sempre que precisarmos.

Uma função pode:

- receber informações;
- realizar cálculos ou outras operações;
- retornar um resultado.

## 2. Criando uma função

Para criar uma função usamos a palavra-chave `def`.

Estrutura básica:

```python
def nome_da_funcao():
    # código da função
```

O código indentado abaixo do `def` pertence à função.

````python
def mostrar_mensagem():
    print("Olá! Esta é uma função.")
````

## 3. Chamando uma função

Criar uma função não significa que ela será executada imediatamente.

Para executar a função, precisamos **chamá-la** pelo seu nome.

````python
mostrar_mensagem()
````

## 4. Funções com parâmetros

Uma função pode receber informações para trabalhar.

Essas informações são chamadas de **parâmetros**.

Exemplo:

````python
def mostrar_nome(nome):
    print("Olá,", nome)
````

Quando chamamos a função, fornecemos um **argumento** para o parâmetro:

````python
mostrar_nome("Ana")
mostrar_nome("Carlos")
````

### Parâmetro x argumento

- **Parâmetro:** variável definida na criação da função.
- **Argumento:** valor fornecido quando a função é chamada.

## 5. Funções com mais de um parâmetro

````python
def somar(a, b):
    print(a + b)
````

````python
somar(10, 5)
somar(20, 30)
````

## 6. Retornando valores com `return`

Até agora nossas funções apenas exibiram informações com `print()`.

Mas muitas vezes precisamos **receber o resultado da função e utilizá-lo em outra parte do programa**.

Para isso usamos `return`.

````python
def somar(a, b):
    resultado = a + b
    return resultado
````

````python
resultado = somar(10, 5)

print(resultado)
````

### `print()` x `return`

`print()` apenas mostra um valor na tela.

`return` devolve um valor para quem chamou a função.

Por isso, quando queremos continuar trabalhando com o resultado, normalmente utilizamos `return`.

````python
def dobro(numero):
    return numero * 2

valor = dobro(8)
resultado = valor + 10

print(resultado)
````

## 7. Funções e cálculos de Engenharia

Funções são muito úteis para organizar cálculos que serão realizados várias vezes.

Por exemplo, podemos criar uma função para calcular a corrente elétrica usando a Lei de Ohm:

$$I = \frac{V}{R}$$

````python
def calcular_corrente(tensao, resistencia):
    corrente = tensao / resistencia
    return corrente

I = calcular_corrente(12, 100)

print("Corrente:", I, "A")
````

## 8. Funções com vários cálculos

Uma função pode realizar várias operações antes de retornar o resultado.

````python
def calcular_potencia(tensao, corrente):
    potencia = tensao * corrente
    return potencia

I = calcular_corrente(12, 100)
P = calcular_potencia(12, I)

print("Corrente:", I, "A")
print("Potência:", P, "W")
````

## 9. Valores padrão

Podemos definir um valor padrão para um parâmetro.

Nesse caso, se o argumento não for informado, o Python utilizará o valor definido na função.

````python
def apresentar_mensagem(nome="Aluno"):
    print("Olá,", nome)
````

````python
apresentar_mensagem()
apresentar_mensagem("Maria")
````

## 10. Variáveis dentro de funções

Uma variável criada dentro de uma função normalmente existe apenas dentro daquela função.

Por isso, é importante diferenciar variáveis locais de variáveis utilizadas fora da função.

````python
def calcular_area(base, altura):
    area = base * altura
    return area

resultado = calcular_area(5, 3)
print(resultado)
````

## 11. Uma função pode chamar outra função

Podemos dividir um problema maior em várias funções menores.

Isso ajuda a deixar o programa mais organizado.

````python
def calcular_corrente(tensao, resistencia):
    return tensao / resistencia

def calcular_potencia(tensao, resistencia):
    corrente = calcular_corrente(tensao, resistencia)
    return tensao * corrente

potencia = calcular_potencia(12, 100)

print("Potência:", potencia, "W")
````

## Resumo

A estrutura mais importante para lembrar é:

```python
def nome_da_funcao(parametro):
    resultado = ...
    return resultado
```

E para utilizar:

```python
valor = nome_da_funcao(argumento)
```

### Conceitos principais

| Conceito | Função |
|---|---|
| `def` | cria uma função |
| parâmetro | recebe uma informação |
| argumento | valor passado para a função |
| `return` | devolve um resultado |
| chamada | executa a função |

A ideia principal é usar funções para **dividir problemas grandes em partes menores e reutilizáveis**.
