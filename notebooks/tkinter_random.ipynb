## 1. Biblioteca `random`

A biblioteca `random` permite gerar valores aleatórios, úteis em simulações, jogos, testes e geração de ruído.

Alguns exemplos:

- `random.random()` → número entre 0 e 1
- `random.randint(a, b)` → inteiro no intervalo
- `random.uniform(a, b)` → float no intervalo
- `random.choice(lista)` → escolha aleatória de uma lista

A aleatoriedade é muito útil em experimentos e quando queremos simular variações reais.

````python
import random

valores = [random.randint(0, 100) for _ in range(10)]
print("Valores gerados:", valores)
print("Maior valor:", max(valores))
print("Menor valor:", min(valores))
print("Valor aleatório entre 0 e 1:", random.random())
````

## 2. Geração de sinal com `numpy` e `matplotlib`

Agora vamos gerar um sinal senoidal e adicionar um pouco de ruído aleatório.

A ideia é simples:

- `np.linspace()` cria valores de tempo;
- `np.sin()` calcula o sinal senoidal;
- `random` ou `np.random` gera variações aleatórias;
- `plt.plot()` mostra o resultado em um gráfico.

Isso é muito comum em sinais, engenharia e ciência de dados.

````python
import numpy as np
import matplotlib.pyplot as plt

amplitude = 2
frequencia = 3
tempo = np.linspace(0, 2, 1000) # gerando um vetor de tempo de 0 a 2 segundos com 1000 pontos
sinal = amplitude * np.sin(2 * np.pi * frequencia * tempo) # gerando um sinal senoidal
ruido = np.random.normal(0, 0.25, len(tempo)) # gerando ruído aleatório com distribuição normal
sinal_com_ruido = sinal + ruido # adicionando o ruído ao sinal original

plt.figure(figsize=(10, 5)) # definindo o tamanho da figura
plt.plot(tempo, sinal, label='Sinal limpo', linewidth=2)
plt.plot(tempo, sinal_com_ruido, label='Sinal com ruído', alpha=0.8)
plt.title('Exemplo de sinal senoidal com ruído aleatório')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
````

## 3. Interface gráfica com `tkinter`

A biblioteca `tkinter` permite criar janelas, botões, campos de texto e outros elementos gráficos.

Neste exemplo, vamos criar uma interface com:

- um campo para inserir a frequência;
- um menu suspenso para escolher o tipo de sinal;
- um botão para gerar um número aleatório;
- um botão para gerar o sinal;
- um gráfico dentro da própria janela.

Os elementos principais que vamos usar são:

- `Entry`: campo de texto para receber valor digitado pelo usuário;
- `ttk.Combobox`: menu suspenso com opções pré-definidas;
- `Button`: botão para executar uma ação.

A estrutura básica é:

```python
import tkinter as tk

root = tk.Tk()
root.title('Minha janela')
root.mainloop()
```

Essa janela fica aberta até o usuário fechar a aplicação.

### Exemplo adicionando uma image com matplotlib

### Por que usar `fig` e `ax` em vez de `plt`?

Quando usamos `plt.plot(...)`, `plt.title(...)`, `plt.xlabel(...)` e etc., estamos usando a interface mais simples do Matplotlib, chamada `pyplot`.

Nesse estilo, o Matplotlib "adivinha" qual é a figura e qual é o eixo atual e aplica a mudança ali. É ótimo para gráficos rápidos, mas pode ficar confuso quando temos mais de um gráfico ou quando estamos construindo interfaces com Tkinter.

Quando usamos `fig = Figure()` e `ax = fig.add_subplot(111)`, estamos trabalhando de forma mais explícita:

- `fig` = a figura inteira (janela do gráfico)
- `ax` = o eixo/área onde o gráfico será desenhado

Por isso os nomes mudam:

- `plt.plot(x, y)` vira `ax.plot(x, y)`
- `plt.title("...")` vira `ax.set_title("...")`
- `plt.xlabel("...")` vira `ax.set_xlabel("...")`
- `plt.grid(True)` vira `ax.grid(True)`

Essa mudança acontece porque agora estamos falando diretamente com o objeto `Axes`, e não com o estado global do Matplotlib. Em projetos maiores, isso deixa o código mais organizado e permite criar vários gráficos em uma mesma janela com mais controle.

Em resumo: `plt` é mais curto e prático; `fig` e `ax` são mais claros, flexíveis e ideais para aplicações com Tkinter, Jupyter e gráficos mais complexos.

````python
import tkinter as tk
import numpy as np

from matplotlib.figure import Figure
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg


# Criar a janela
root = tk.Tk()
root.title("Gráfico de Seno")
root.geometry("700x500")


# Criar os valores de x
x = np.linspace(0, 2 * np.pi, 100)

# Calcular o seno
y = np.sin(x)


# Criar o gráfico
fig = Figure(figsize=(6, 4))

# Adicionar um subplot na posição 111 (1 linha, 1 coluna, 1º gráfico)
ax = fig.add_subplot(111)

# Plota o gráfico de seno
ax.plot(x, y)

# Adicionar título e rótulos aos eixos
ax.set_title("Função Seno")
# Adicionar rótulos aos eixos
ax.set_xlabel("x")
ax.set_ylabel("sen(x)")
# Adicionar grade ao gráfico
ax.grid(True)


# Colocar o gráfico dentro da janela
canvas = FigureCanvasTkAgg(fig, master=root)

# Renderizar o gráfico
canvas.get_tk_widget().pack()


# Manter a janela aberta
root.mainloop()
````

### Exemplo com matplotlib e funções

````python
import tkinter as tk
from tkinter import ttk
import numpy as np
import matplotlib
matplotlib.use('TkAgg')
from matplotlib.figure import Figure
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

root = tk.Tk()
root.title('Tkinter + random + sinal')
root.geometry('700x500')

label = tk.Label(root, text='Clique no botão para gerar um número aleatório', font=('Arial', 12))
label.pack(pady=10)

fig = Figure(figsize=(6, 3), dpi=100)
ax = fig.add_subplot(111)
canvas = FigureCanvasTkAgg(fig, master=root)
canvas.get_tk_widget().pack(pady=10)


def gerar_sinal():
    frequencia = 4
    amplitude = 1.5
    tempo = np.linspace(0, 2, 500)
    sinal = amplitude * np.sin(2 * np.pi * frequencia * tempo)
    ruido = np.random.normal(0, 0.2, len(tempo))
    sinal_ruidoso = sinal + ruido

    plotar_sinal(tempo, sinal, sinal_ruidoso)


def plotar_sinal(tempo, sinal, sinal_ruidoso):
    ax.clear()
    ax.plot(tempo, sinal, label='Sinal limpo', color='blue')
    ax.plot(tempo, sinal_ruidoso, label='Com ruído', color='red', alpha=0.8)
    ax.set_title('Sinal gerado')
    ax.set_xlabel('Tempo (s)')
    ax.set_ylabel('Amplitude')
    ax.legend()
    ax.grid(True)
    canvas.draw()
    label.config(text='Sinal plotado com sucesso!')

botao_sinal = ttk.Button(root, text='Gerar sinal', command=gerar_sinal)
botao_sinal.pack(pady=5)

root.mainloop()
````

````python
import tkinter as tk
from tkinter import ttk
import numpy as np
import matplotlib
matplotlib.use('TkAgg')
from matplotlib.figure import Figure
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

root = tk.Tk()
root.title('Exemplo com Entry e ComboBox')
root.geometry('800x550')

# Campo para inserir a label frequência
label_frequencia = tk.Label(root, text='Frequência (Hz):', font=('Arial', 11))
label_frequencia.pack(pady=(10, 0))

# Campo de entrada do valor da frequência
entrada = tk.Entry(root, width=20, font=('Arial', 12))
entrada.insert(0, '4')
entrada.pack(pady=5)

# Menu suspenso
label_tipo = tk.Label(root, text='Tipo de sinal:', font=('Arial', 11))
label_tipo.pack(pady=(10, 0))

tipo_sinal = tk.StringVar(value='Seno')
combo = ttk.Combobox(root, textvariable=tipo_sinal, values=['Seno', 'Quadrada', 'Dente de serra'], state='readonly', width=20)
combo.pack(pady=5)

# Exibe mensagens na tela
status = tk.StringVar(value='Aguardando entrada...')
label_status = tk.Label(root, textvariable=status, font=('Arial', 12, 'bold'))
label_status.pack(pady=10)

# Gráfico
fig = Figure(figsize=(7, 3.5), dpi=100)
ax = fig.add_subplot(111)
canvas = FigureCanvasTkAgg(fig, master=root)
canvas.get_tk_widget().pack(pady=10)

def gerar_sinal():
    try:
        frequencia = float(entrada.get())
    except ValueError:
        status.set('Digite uma frequência válida!')
        return

    tipo = combo.get()
    tempo = np.linspace(0, 2, 500)

    if tipo == 'Seno':
        sinal = np.sin(2 * np.pi * frequencia * tempo)
    elif tipo == 'Quadrada':
        sinal = np.sign(np.sin(2 * np.pi * frequencia * tempo))
    else:
        sinal = 2 * (frequencia * tempo % 1) - 1

    ax.clear()
    ax.plot(tempo, sinal, label=tipo, color='blue', linewidth=2)
    ax.set_title(f'Sinal {tipo}')
    ax.set_xlabel('Tempo (s)')
    ax.set_ylabel('Amplitude')
    ax.legend()
    ax.grid(True)
    canvas.draw()
    status.set(f'Frequência: {frequencia} Hz | Tipo: {tipo}')


botao_sinal = ttk.Button(root, text='Gerar sinal', command=gerar_sinal)
botao_sinal.pack(pady=5)

root.mainloop()
````

````python
import tkinter as tk
from tkinter import ttk
import numpy as np
import matplotlib
matplotlib.use('TkAgg')
from matplotlib.figure import Figure
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg


root = tk.Tk()
root.title('Gerador de sinal')
root.geometry('700x500')

label_freq = tk.Label(root, text='Frequência (Hz):')
label_freq.pack(pady=(10, 0))

entrada = tk.Entry(root)
entrada.insert(0, '5')
entrada.pack(pady=5)

fig = Figure(figsize=(6, 3), dpi=100)
ax = fig.add_subplot(111)
canvas = FigureCanvasTkAgg(fig, master=root)
canvas.get_tk_widget().pack(pady=10)


def gerar_sinal_gui():
    frequencia = float(entrada.get())
    tempo = np.linspace(0, 2, 500)
    sinal = np.sin(2 * np.pi * frequencia * tempo)

    ax.clear()
    ax.plot(tempo, sinal, color='green', linewidth=2)
    ax.set_title(f'Sinal com {frequencia} Hz')
    ax.set_xlabel('Tempo (s)')
    ax.set_ylabel('Amplitude')
    ax.grid(True)
    canvas.draw()


botao = ttk.Button(root, text='Gerar sinal', command=gerar_sinal_gui)
botao.pack(pady=10)

root.mainloop()
````
